```python
data_salary = response_salary.json()

years = []
sexes = []
salaries = []

for item in data_salary['data']:
    years.append(int(item['key'][2]))
    sexes.append('Men' if item['key'][1] == '1' else 'Women')
    salaries.append(int(item['values'][0]))

df_salary = pd.DataFrame({
    'year': years,
    'sex': sexes,
    'monthly_salary': salaries
})

print(df_salary)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# --- Sol grafik: 2023 karşılaştırması ---
df_2023 = df_salary[df_salary['year'] == 2023]
axes[0].bar(df_2023['sex'], df_2023['monthly_salary'], 
            color=['steelblue', 'coral'], width=0.5)
axes[0].set_title('Average Monthly Salary 2023', fontsize=13, fontweight='bold')
axes[0].set_ylabel('Monthly Salary (SEK)')
axes[0].set_ylim(0, 50000)

for i, row in df_2023.iterrows():
    axes[0].text(row['sex'], row['monthly_salary'] + 500, 
                f"{row['monthly_salary']:,} SEK", 
                ha='center', fontweight='bold')

# --- Sağ grafik: 2024 karşılaştırması ---
df_2024 = df_salary[df_salary['year'] == 2024]
axes[1].bar(df_2024['sex'], df_2024['monthly_salary'],
            color=['steelblue', 'coral'], width=0.5)
axes[1].set_title('Average Monthly Salary 2024', fontsize=13, fontweight='bold')
axes[1].set_ylabel('Monthly Salary (SEK)')
axes[1].set_ylim(0, 50000)

for i, row in df_2024.iterrows():
    axes[1].text(row['sex'], row['monthly_salary'] + 500,
                f"{row['monthly_salary']:,} SEK",
                ha='center', fontweight='bold')

plt.suptitle('Sweden Gender Pay Gap Analysis', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.savefig('gender_pay_gap.png', dpi=150)
plt.show()
```

```python
men_2023 = df_salary[(df_salary['year'] == 2023) & (df_salary['sex'] == 'Men')]['monthly_salary'].values[0]
men_2024 = df_salary[(df_salary['year'] == 2024) & (df_salary['sex'] == 'Men')]['monthly_salary'].values[0]

women_2023 = df_salary[(df_salary['year'] == 2023) & (df_salary['sex'] == 'Women')]['monthly_salary'].values[0]
women_2024 = df_salary[(df_salary['year'] == 2024) & (df_salary['sex'] == 'Women')]['monthly_salary'].values[0]

gap_2023 = ((men_2023 - women_2023) / men_2023) * 100
gap_2024 = ((men_2024 - women_2024) / men_2024) * 100

print("=" * 40)
print("📊 GENDER PAY GAP SUMMARY")
print("=" * 40)
print(f"2023 gap: {gap_2023:.1f}%")
print(f"2024 gap: {gap_2024:.1f}%")
print(f"Fark: {gap_2024 - gap_2023:.1f}%")
```
